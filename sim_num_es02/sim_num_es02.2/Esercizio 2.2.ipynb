```python
import matplotlib.pyplot as plt
import numpy as np

data1 = np.loadtxt("continuous.dat")
pass1 = data1[:, 0]
int1 = data1[:, 1]
err1 = data1[:, 2]

data2 = np.loadtxt("lattice.dat")
pass2 = data1[:, 0]
int2 = data1[:, 1]
err2 = data1[:, 2]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(pass1, int1, 'o-', markersize=3)
ax1.set_xlabel("Step")
ax1.set_ylabel(r"$\sqrt{\langle |\vec{r}_N|^2 \rangle_{RW}}$")
ax1.grid(True, alpha=0.3)

ax2.plot(pass2, int2, 'o-', markersize=3)
ax2.set_xlabel("Step")
ax2.set_ylabel(r"$\sqrt{\langle |\vec{r}_N|^2 \rangle_{RW}}$")
ax2.grid(True, alpha=0.3)

fig.suptitle('Ciao', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

step = 10

# curva 1: continuous
plt.plot(pass1, int1, label="continuous", color="C0")
plt.errorbar(pass1[::step], int1[::step], yerr=10*err1[::step],
             fmt='o', capsize=5, color="C0")

# curva 2: lattice
plt.plot(pass2, int2, label="lattice", color="C1")
plt.errorbar(pass2[::step], int2[::step], yerr=10*err2[::step],
             fmt='o', capsize=5, color="C1")

plt.xlabel("step i")
plt.ylabel(r"$\sqrt{\langle |\vec{r}_N|^2 \rangle_{RW}}$")

plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
from scipy.optimize import curve_fit

def rw_model(N, a):
    return a * np.sqrt(N)

popt, pcov = curve_fit(rw_model, pass1, int1)

a_fit = popt[0]
a_err = np.sqrt(pcov[0,0])

print("a =", a_fit, "+/-", a_err)
```

```python
plt.figure(figsize=(8,5))

plt.errorbar(pass1, int1, yerr=err1, fmt='o', markersize=4, label="dati")
plt.plot(pass1, rw_model(pass1, a_fit), label=f"fit: a={a_fit:.3f}", color="red")

plt.xlabel("N (step)")
plt.ylabel(r"$\sqrt{\langle r_N^2 \rangle}$")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
