```python
import matplotlib.pyplot as plt
import numpy as np

data1 = np.loadtxt("uniform.dat")
int1 = data1[:, 0]
err1 = data1[:, 1]
pass1 = data1[:, 2]

data2 = np.loadtxt("non_uniform.dat")
int2 = data2[:, 0]
err2 = data2[:, 1]
pass2 = data2[:, 2]

step = 500

idx1 = np.arange(0, len(pass1), step)
idx2 = np.arange(0, len(pass2), step)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(pass1, int1, 'o-', markersize=3, label="Uniform")
ax1.errorbar(pass1[idx1], int1[idx1], yerr=err1[idx1],
             fmt='o', capsize=5, color='red')

ax1.set_xlabel("Passi")
ax1.set_ylabel("Integrale")
ax1.grid(True, alpha=0.3)

ax2.plot(pass2, int2, 'o-', markersize=3, label="Importance")
ax2.errorbar(pass2[idx2], int2[idx2], yerr=err2[idx2],
             fmt='o', capsize=5, color='red')

ax2.set_xlabel("Passi")
ax2.set_ylabel("Integrale")
ax2.grid(True, alpha=0.3)

fig.suptitle('Confronto tra valori integrale (uniforme vs importance)', fontsize=10)

plt.tight_layout()
plt.show()

```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(pass1, err1, 'o-')
ax1.set_xlabel("Passi")
ax1.set_ylabel("Errore")
ax1.grid(True, alpha=0.3)

ax2.plot(pass2, err2, 'o-')
ax2.set_xlabel("Passi")
ax2.set_ylabel("Errore")
ax2.grid(True, alpha=0.3)

fig.suptitle('Confronto tra errori con distribuzione uniforme e importance sampling', fontsize=10)

plt.tight_layout()
plt.show()
```
