# Esercizio 8.1

Prima di implementare il Simulated Annealing, ho effettuato una scansione dello spazio dei parametri $\sigma$ e $\mu$, in modo da individuare la regione in cui si trova il minimo dell'energia e avere quindi un'idea dei valori ottimali da cui partire.

In particolare, considero i due parametri nell'intervallo

$$
0.5 \leq \sigma \leq 1.0,
\qquad
0.5 \leq \mu \leq 1.0,
$$

con passo $0.05$. Per ogni coppia $(\sigma,\mu)$ eseguo una simulazione Monte Carlo indipendente, utilizzando l'algoritmo di Metropolis per campionare la distribuzione $|\psi(x)|^2$.

La funzione d'onda utilizzata è

$$
\psi(x)=
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
+
e^{-\frac{(x+\mu)^2}{2\sigma^2}},
$$

e quindi la probabilità di accettazione di una nuova posizione $x_{\mathrm{new}}$ è

$$
A=\min\left(1,
\frac{|\psi(x_{\mathrm{new}})|^2}
{|\psi(x)|^2}
\right).
$$

Per ogni configurazione dei parametri effettuo inizialmente una fase di equilibrazione di $5000$ passi, durante la quale il sistema viene portato nella distribuzione di equilibrio. Successivamente eseguo $N=100000$ lanci, suddivisi in $M=1000$ blocchi da $n=100$ passi ciascuno.

All'interno di ogni blocco calcolo il valor medio dell'energia locale

$$
E_{\mathrm{loc}}(x)
=
-\frac{1}{2}
\frac{\nabla^2\psi(x)}{\psi(x)}
+x^4-\frac{5}{2}x^2.
$$

I valori ottenuti vengono poi elaborati tramite il metodo del data blocking, ottenendo per ogni blocco il valor medio progressivo dell'energia e la relativa incertezza statistica.

Infine, confronto i valori finali dell'energia ottenuti per tutte le configurazioni e memorizzo quella corrispondente al minimo globale trovato. In particolare, salvo i valori

$$
E_{\min}, \qquad \sigma_{\mathrm{best}}, \qquad \mu_{\mathrm{best}},
$$

che rappresentano rispettivamente la minima energia trovata nella scansione e i valori dei parametri a cui essa corrisponde.

Questa scansione permette quindi di avere una prima stima della posizione del minimo nello spazio dei parametri $(\sigma,\mu)$ e costituisce un controllo preliminare dei risultati che verranno successivamente ottenuti tramite il Simulated Annealing.


# Esercizio 8.2

Per questo esercizio ho bisogno di implementare un algoritmo di Simulated Annealing per ottimizzare i due parametri variabili della funzione d’onda, ovvero $\sigma$ e $\mu$, al fine di trovare la configurazione che minimizza l’energia dello stato fondamentale. Per calcolare l’energia associata a una determinata coppia di parametri utilizzo un algoritmo di Metropolis VMC, analogo a quello utilizzato nell’esercizio precedente, che permette di campionare le coordinate $x$ secondo la distribuzione $|\psi(x)|^2$.

Prima di tutto, ho implementato alcune funzioni necessarie per il calcolo dell’energia. Le funzioni `psi` e `psi2` restituiscono rispettivamente la funzione d’onda $\psi(x)$ e il suo modulo quadro $|\psi(x)|^2$. La funzione `nabla2_psi` calcola invece la derivata seconda della funzione d’onda, necessaria per determinare l’energia locale. Infine, la funzione `E_loc` restituisce il valore dell’energia locale

$$
E_{\mathrm{loc}}(x)=-\frac{1}{2}\frac{\nabla^2\psi(x)}{\psi(x)}+V(x),
$$

dove il potenziale considerato è

$$
V(x)=x^4-\frac{5}{2}x^2.
$$

Ho inoltre definito una struttura `result`, contenente sia il valore medio dell’energia sia la relativa incertezza statistica, in modo da poter ottenere entrambi i risultati dalla funzione `VMC`.

### 1. Funzione `VMC`

La funzione `VMC` implementa il campionamento delle coordinate $x$ tramite l'algoritmo di Metropolis. Prende in ingresso il generatore di numeri casuali, i parametri $\sigma$ e $\mu$, il numero totale di lanci e un parametro booleano che indica se salvare o meno i dati.

Per ogni configurazione propongo una nuova posizione secondo

$$
x_{\mathrm{new}}=x+\delta,
$$

dove $\delta$ è estratto uniformemente nell'intervallo $[-\Delta,\Delta]$, con $\Delta=1$. La probabilità di accettazione è

$$
A=\min\left(1,\frac{|\psi(x_{\mathrm{new}})|^2}{|\psi(x)|^2}\right).
$$

Dopo una fase iniziale di equilibrazione di 5000 passi, eseguo il campionamento vero e proprio. Ho scelto di dividere i \(N=100000\) lanci in \(M=N/1000\) blocchi, ciascuno contenente 1000 estrazioni. Per ogni blocco calcolo il valor medio dell'energia locale e successivamente applico il data blocking per ottenere il valor medio progressivo e la relativa incertezza.

La funzione restituisce infine il valore medio dell'energia ottenuto nell'ultimo blocco progressivo e il relativo errore statistico.

Quando richiesto, salvo inoltre le posizioni campionate nel file `positions.dat` e i valori progressivi di energia ed errore nel file `final_energy.dat`.

### 2. Implementazione del Simulated Annealing

Nel `main` implemento direttamente l'algoritmo di Simulated Annealing. Parto dalla configurazione iniziale

$$
\sigma=0.8,\qquad \mu=0.8
$$

e calcolo la corrispondente energia tramite la funzione `VMC`. Questa configurazione viene inizialmente considerata come configurazione corrente.

Introduco quindi una temperatura iniziale

$$
T=1
$$

e una temperatura minima

$$
T_{\min}=0.01.
$$

Ad ogni iterazione dell'algoritmo propongo una nuova configurazione nello spazio dei parametri, modificando indipendentemente $\sigma$ e $\mu$:

$$
\sigma_{\mathrm{new}}=\sigma+\Delta\sigma,
\qquad
\mu_{\mathrm{new}}=\mu+\Delta\mu,
$$

dove gli spostamenti sono estratti uniformemente nell'intervallo $[-0.05,0.05]$. Poiché $\sigma$ rappresenta una deviazione standard, non può assumere valori negativi; per questo motivo, se il nuovo valore di $\sigma$ risulta minore o uguale a zero, la configurazione viene scartata e viene proposta una nuova mossa.

Per ogni nuova coppia $(\sigma_{\mathrm{new}},\mu_{\mathrm{new}})$, calcolo l'energia tramite una nuova simulazione VMC. Indico con $E_{\mathrm{old}}$ l'energia della configurazione corrente e con $E_{\mathrm{new}}$ quella della configurazione proposta.

Calcolo quindi la differenza di energia

$$
\Delta E=E_{\mathrm{new}}-E_{\mathrm{old}}.
$$

Se $\Delta E<0$, la nuova configurazione ha energia minore e viene quindi sempre accettata. Se invece $\Delta E>0$, la configurazione viene accettata con probabilità

$$
P=e^{-\Delta E/T}.
$$

In questo modo è possibile accettare occasionalmente configurazioni energeticamente peggiori, evitando che l'algoritmo rimanga bloccato in un minimo locale. Quando una configurazione viene accettata, aggiorno i valori correnti di $\sigma$, $\mu$ ed energia.


### 3. Legge di raffreddamento

Dopo ogni iterazione aggiorno la temperatura utilizzando una legge di raffreddamento geometrica:

$$
T_{\mathrm{new}}=0.98\,T.
$$

Partendo da $T=1$, questa legge permette di diminuire progressivamente la probabilità di accettare configurazioni con energia maggiore. Il sistema passa quindi gradualmente da una fase in cui può esplorare liberamente lo spazio dei parametri a una fase in cui tende a rimanere vicino alle configurazioni a energia minima.

Il ciclo termina quando la temperatura raggiunge il valore $T_{\min}=0.01$.

Durante il Simulated Annealing salvo nel file `SA.dat` la temperatura, i valori correnti di $\sigma$ e $\mu$ l'energia corrente e il relativo errore. Questi dati possono essere utilizzati per analizzare l'evoluzione dell'algoritmo e verificare la convergenza verso la configurazione ottimale.

### 4. Calcolo finale dell'energia

Al termine del Simulated Annealing considero come configurazione ottimale quella caratterizzata dai valori `sigma_best` e `mu_best`, corrispondenti alla minima energia incontrata durante l'intera simulazione.

Utilizzo quindi nuovamente la funzione `VMC` con questi parametri, questa volta impostando `save_data=true`. In questo modo eseguo una simulazione finale indipendente per ottenere la migliore stima dell'energia del ground state e salvo le posizioni $x$ campionate nel file `positions.dat`.

Le configurazioni campionate possono infine essere utilizzate per costruire l'istogramma della distribuzione di probabilità $
|\psi(x)|^2,
$ permettendo di confrontare la distribuzione ottenuta numericamente con quella attesa per la funzione d'onda ottimizzata.


```python
data = np.loadtxt('sim_num_es08.2/SA.dat')
error = data[:, 4]
energy = data[:, 3]
steps = np.arange(len(energy))

plt.errorbar(steps, energy, yerr=error, fmt='.')
plt.plot(energy)

plt.xlabel("SA step")
plt.ylabel(r"$\langle H\rangle_T$")
plt.grid()
plt.show()
```

## Commento
Il grafico mostra l’evoluzione dell’energia variazionale $\langle H\rangle_T$ al variare dei passi del
Simulated Annealing. Si osserva una chiara tendenza alla diminuzione dell’energia
man mano che la temperatura viene abbassata, con fluttuazioni più ampie nelle fasi
iniziali ($T$ alta) e una progressiva stabilizzazione quando il sistema esplora regioni
più favorevoli dello spazio dei parametri.

Le barre d’errore rappresentano l’incertezza statistica associata alla stima
dell’energia in ciascun passo SA: la riduzione complessiva di $\langle H\rangle_T$ è significativamente
più grande delle fluttuazioni, indicando che l’algoritmo ha effettivamente trovato
configurazioni variazionali più favorevoli.

```python
data = np.loadtxt('sim_num_es08.2/SA.dat')
sigma = data[:, 1]
mu = data[:, 2]
steps = np.arange(len(mu))

plt.scatter(sigma, mu, c=steps)

plt.xlabel(r'$\sigma$')
plt.ylabel(r'$\mu$')
plt.title('SA trajectory in parameter space')
plt.colorbar(label='SA step')
plt.grid()
plt.show()
```

```python
data = np.loadtxt("sim_num_es08.2/final_energy.dat")

blocks = data[:, 0]
energy = data[:, 1]
error = data[:, 2]

plt.errorbar(
    blocks,
    energy,
    yerr=error,
    fmt='.'
)

plt.xlabel("Number of blocks")
plt.ylabel(r"$\langle H\rangle_T$")
plt.title("VMC energy convergence")
plt.grid()
plt.show()
```

## Commento

Il grafico mostra la stima dell’energia variazionale $\langle H\rangle_T$ ottenuta tramite data
blocking, utilizzando i parametri $(\simga, \mu)$ che minimizzano l’energia durante il SA.
La media cumulativa converge rapidamente verso un valore stabile, mentre
l’incertezza statistica diminuisce come $1/\sqrt(N)$, come previsto dal Teorema del limite Centrale.

La stabilizzazione dell’energia e la riduzione dell’errore indicano che il numero di
campioni utilizzato è sufficiente per una stima affidabile dell’energia variazionale
associata alla funzione d’onda ottimizzata.

```python
x = np.loadtxt("sim_num_es08.2/positions.dat")
plt.hist(x, bins=100, density=True)

N = 500
x_min, x_max = -3, 3
x = np.linspace(x_min, x_max, N)
dx = x[1] - x[0]
V = x**4 - 2.5*x**2
H = np.zeros((N, N))
for i in range(N):
    H[i, i] = 1/dx**2 + V[i]
    if i > 0:
        H[i, i-1] = -0.5/dx**2
    if i < N-1:
        H[i, i+1] = -0.5/dx**2
E, psi = np.linalg.eigh(H)
psi0 = psi[:, 0] 
psi0 /= np.sqrt(np.trapz(psi0**2, x))
plt.plot(x, psi0**2, label="Numerical GS")

sigma = 0.626664
mu = 0.821381
xx = np.linspace(-3, 3, 1000)
psi = np.exp(-(xx-mu)**2/(2*sigma**2)) \
    + np.exp(-(xx+mu)**2/(2*sigma**2))
psi2 = psi**2
psi2 /= np.trapz(psi2, xx)
plt.plot(xx, psi2, label=r"$|\Psi_T(x)|^2$")


plt.xlabel(r"$x$")
plt.ylabel(r"$|\Psi_T(x)|^2$")
plt.title("Metropolis sampling")
plt.grid()
plt.show()
```

## Commento

L’istogramma mostra la distribuzione delle posizioni campionate tramite algoritmo
Metropolis, che dovrebbe riprodurre $|\psi_t(x)|$. La forma a doppio picco è chiaramente
visibile e coincide con la struttura della funzione d’onda trial ottimizzata.

Il confronto con la curva analitica di $|\psi_t(x)|$ e con la soluzione numerica ottenuta
dalla discretizzazione dell’equazione di Schrödinger conferma che il campionamento
Metropolis è corretto: le regioni di maggiore densità di probabilità coincidono con
quelle previste teoricamente, e la funzione trial ottimizzata fornisce una buona
approssimazione dello stato fondamentale.
