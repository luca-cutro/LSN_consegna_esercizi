```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def read_value(filename):
    data = np.loadtxt(filename)
    
    ave = data[:, 0] # legge sum_ave che è il valor medio dei blocchi
    err = data[:, 1] # legge l'errore
    
    return ave, err # da l'ultima riga
```

```python
Block, T = read_value("OUTPUT/temperature.dat")

plt.figure(figsize=(10, 6))
plt.scatter(Block, T, s=20, alpha=0.6)
plt.xlabel("Block")
plt.ylabel("Temperature")
plt.grid()
plt.show()
```

```python
Block, U = read_value("OUTPUT/potential_energy.dat")

plt.figure(figsize=(10, 6))
plt.scatter(Block, U, s=20, alpha=0.6)
plt.xlabel("Block")
plt.ylabel("U / N")
plt.grid()
plt.show()
```

```python
def read_value(filename):
    data = np.loadtxt(filename)
    
    ave = data[:, 1] # legge sum_ave che è il valor medio dei blocchi
    err = data[:, 2] # legge l'errore
    acc = data[:, 3]
    
    return ave, err, acc # da l'ultima riga
```
