## Gibbs

### Algoritmo di Gibbs

L'algoritmo di Gibbs è una tecnica di campionamento Monte Carlo basata sulla simulazione di una catena di Markov. Come nel caso dell'algoritmo di Metropolis, l'obiettivo è generare configurazioni distribuite secondo la distribuzione di Boltzmann, ma con una modalità diversa di aggiornamento.

Nel modello di Ising ogni particella è rappresentata da uno spin

$$
s_i=\pm1,
$$

e l'energia del sistema è

$$
E=-J\sum_{\langle i,j\rangle}s_i s_j.
$$

L'algoritmo di Gibbs aggiorna gli spin uno alla volta. Scelto uno spin $s_i$, si considerano fissi tutti gli altri spin e si calcola la probabilità condizionata che $s_i$ assuma il valore $+1$ oppure $-1$.

Definendo il campo locale generato dagli spin vicini come

$$
h_i=J\sum_{j\in nn(i)}s_j,
$$

la probabilità condizionata di avere $s_i=+1$ è

$$
P(s_i=+1|\{s_j\}_{j\neq i})
=
\frac{e^{\beta h_i}}
{e^{\beta h_i}+e^{-\beta h_i}}
=
\frac{1}{1+e^{-2\beta h_i}},
$$

mentre

$$
P(s_i=-1)=1-P(s_i=+1).
$$

Si genera quindi un numero casuale uniforme $r\in[0,1)$. Se

$$
r<P(s_i=+1),
$$

si assegna $s_i=+1$, altrimenti si assegna $s_i=-1$.

La differenza principale rispetto all'algoritmo di Metropolis è che non viene proposta una configurazione per poi decidere se accettarla o rifiutarla. Nel metodo di Gibbs il nuovo valore dello spin viene invece estratto direttamente dalla sua distribuzione di probabilità condizionata. La mossa viene quindi sempre effettuata.

Ripetendo il procedimento per tutti gli spin e per un numero sufficientemente elevato di iterazioni, la catena di Markov converge alla distribuzione di Boltzmann. Dopo il tempo di equilibrazione, le configurazioni generate possono quindi essere utilizzate per calcolare le proprietà termodinamiche del modello.

In particolare, nel nostro studio utilizziamo l'algoritmo di Gibbs per confrontare il suo comportamento con quello dell'algoritmo di Metropolis e verificare che entrambi conducano alla stessa distribuzione di equilibrio, pur utilizzando procedure di aggiornamento differenti.


```python
def read_last_block_Energy(filename):
    data = np.loadtxt(filename)

    T = data[:, 0]
    block = data[:, 2]
    ave = data[:, 3]
    err = data[:, 4]

    T_list = []
    ave_list = []
    err_list = []

    # Temperature presenti nel file
    temperatures = np.unique(T)

    for temp in temperatures:

        # Prendo tutte le righe di questa temperatura
        mask = (T == temp)

        # Ultima riga = ultimo blocco
        ave_temp = ave[mask][-1]
        err_temp = err[mask][-1]

        T_list.append(temp)
        ave_list.append(ave_temp)
        err_list.append(err_temp)

    return np.array(T_list), np.array(ave_list), np.array(err_list)

def read_last_block(filename):
    data = np.loadtxt(filename)

    T = data[:, 0]
    block = data[:, 1]
    ave = data[:, 2]
    err = data[:, 3]

    T_list = []
    ave_list = []
    err_list = []

    # Temperature presenti nel file
    temperatures = np.unique(T)

    for temp in temperatures:

        # Prendo tutte le righe di questa temperatura
        mask = (T == temp)

        # Ultima riga = ultimo blocco
        ave_temp = ave[mask][-1]
        err_temp = err[mask][-1]

        T_list.append(temp)
        ave_list.append(ave_temp)
        err_list.append(err_temp)

    return np.array(T_list), np.array(ave_list), np.array(err_list)
```

## Magnetization

```python
T, M_sim, M_err = read_last_block("sim_num_es06.1/OUTPUT/magnetization.dat")

h = 0.02  # external field
b = 1/T

l1 = np.exp(b*J)*np.cosh(b*h) + np.sqrt(np.exp(2*b*J)*np.cosh(b*h)**2 - 2*np.sinh(2*b*J))
l2 = np.exp(b*J)*np.cosh(b*h) - np.sqrt(np.exp(2*b*J)*np.cosh(b*h)**2 - 2*np.sinh(2*b*J))

Z = l1**Ns + l2**Ns

M_theory = (
    np.exp(b*J)*np.sinh(b*h) *
    (
        l1**(Ns-1) * (1 + np.exp(b*J)*np.cosh(b*h) / np.sqrt(np.exp(2*b*J)*np.cosh(b*h)**2 - 2*np.sinh(2*b*J)))
        + l2**(Ns-1) * (1 - np.exp(b*J)*np.cosh(b*h) / np.sqrt(np.exp(2*b*J)*np.cosh(b*h)**2 - 2*np.sinh(2*b*J)))
    )
) / Z

diff = M_sim - M_theory

fig, ax = plt.subplots(1, 2, figsize=(14, 5))

ax[0].plot(T, M_theory, label="Teorica", color="black")
ax[0].errorbar(T, M_sim, yerr=M_err, fmt='o', label="Simulata", color="blue")
ax[0].set_xlabel("Temperature")
ax[0].set_ylabel("Magnetization")
ax[0].set_title("Magnetizzazione: simulata vs teorica")
ax[0].grid()
ax[0].legend()

ax[1].errorbar(T, diff, yerr=M_err, fmt='o', color="red")
ax[1].axhline(0, color="black", linestyle="--")
ax[1].set_xlabel("Temperature")
ax[1].set_ylabel("M_sim - M_theory")
ax[1].set_title("Differenza simulazione - teoria")
ax[1].grid()

plt.tight_layout()
plt.show()

```

## Total Energy

```python
T_U, U_sim, U_err = read_last_block_Energy("sim_num_es06.1/OUTPUT/total_energy.dat")

beta_U = 1 / T_U
th_U   = np.tanh(beta_U * J)
ch_U   = np.cosh(beta_U * h)
thN_U  = th_U**Ns

e_theory = -J * (th_U + ch_U * thN_U) / (1 + thN_U)

diff = U_sim - e_theory

fig, ax = plt.subplots(1, 2, figsize=(14, 5))

ax[0].errorbar(T_U, U_sim, yerr=U_err, fmt='o', label="Simulata")
ax[0].plot(T_U, e_theory, label="Teorica", color="black")
ax[0].set_xlabel("Temperature")
ax[0].set_ylabel("Total Energy")
ax[0].set_title("Energia totale: simulata vs teorica")
ax[0].grid()
ax[0].legend()

ax[1].errorbar(T_U, diff, yerr=U_err, fmt='o', color="red")
ax[1].axhline(0, color="black", linestyle="--")
ax[1].set_xlabel("Temperature")
ax[1].set_ylabel("U_sim - U_theory")
ax[1].set_title("Differenza simulazione - teoria")
ax[1].grid()

plt.tight_layout()
plt.show()
```

## Susceptibility

```python
T_Chi, Chi_sim, Chi_err = read_last_block("sim_num_es06.1/OUTPUT/susceptibility.dat")

beta_Chi = 1 / T_Chi
th_Chi   = np.tanh(beta_Chi * J)
ch_Chi   = np.cosh(beta_Chi * h)
thN_Chi  = th_Chi**Ns
Chi_theory = beta_Chi * np.exp(2*beta_Chi*J) * (1 - thN_Chi) / (1 + thN_Chi)

diff = Chi_sim - Chi_theory

fig, ax = plt.subplots(1, 2, figsize=(14, 5))

ax[0].errorbar(T_Chi, Chi_sim, yerr=Chi_err, fmt='o', label="Simulata")
ax[0].plot(T_Chi, Chi_theory, label="Teorica", color="black")
ax[0].set_xlabel("Temperature")
ax[0].set_ylabel("Susceptibility")
ax[0].set_title("Suscettibilità: simulata vs teorica")
ax[0].grid()
ax[0].legend()

ax[1].errorbar(T_Chi, diff, yerr=Chi_err, fmt='o', color="red")
ax[1].axhline(0, color="black", linestyle="--")
ax[1].set_xlabel("Temperature")
ax[1].set_ylabel("Chi_sim - Chi_theory")
ax[1].set_title("Differenza simulazione - teoria")
ax[1].grid()

plt.tight_layout()
plt.show()
```

## Specific Heat

```python
T_Cv, Cv_sim, Cv_err = read_last_block("sim_num_es06.1/OUTPUT/specific_heat.dat")

beta_Cv = 1 / T_Cv
th_Cv   = np.tanh(beta_Cv * J)
ch_Cv   = np.cosh(beta_Cv * h)
thN_Cv  = th_Cv**Ns

Cv_theory = ((beta_Cv * J)**2) * (
    ((1 + thN_Cv + (Ns-1)*(th_Cv**2) + (Ns-1)*(ch_Cv**2)*thN_Cv) / (1 + thN_Cv))
    - Ns * ((th_Cv + ch_Cv * thN_Cv) / (1 + thN_Cv))**2
)

diff = Cv_sim - Cv_theory

fig, ax = plt.subplots(1, 2, figsize=(14, 5))

ax[0].errorbar(T_Cv, Cv_sim, yerr=Cv_err, fmt='o', label="Simulata")
ax[0].plot(T_Cv, Cv_theory, label="Teorica", color="black")
ax[0].set_xlabel("Temperature")
ax[0].set_ylabel("Specific Heat")
ax[0].set_title("Calore specifico: simulato vs teorico")
ax[0].grid()
ax[0].legend()

ax[1].errorbar(T_Cv, diff, yerr=Cv_err, fmt='o', color="red")
ax[1].axhline(0, color="black", linestyle="--")
ax[1].set_xlabel("Temperature")
ax[1].set_ylabel("Cv_sim - Cv_theory")
ax[1].set_title("Differenza simulazione - teoria")
ax[1].grid()

plt.tight_layout()
plt.show()
```

## Metropolis

### Algoritmo di Metropolis

L'algoritmo di Metropolis è una tecnica di campionamento Monte Carlo basata sulla simulazione di una catena di Markov. L'obiettivo è generare una sequenza di configurazioni distribuite secondo una determinata distribuzione di probabilità target, senza doverne conoscere esplicitamente la costante di normalizzazione.

A partire dalla configurazione corrente $x$, viene proposta una nuova configurazione candidata $x'$, estratta da una distribuzione di prova $q(x'|x)$. Nel caso di una distribuzione di prova simmetrica, $q(x'|x)=q(x|x')$, la nuova configurazione viene accettata con probabilità

$$
A(x\rightarrow x')=
\min\left[1,\frac{p(x')}{p(x)}\right].
$$

Si genera quindi un numero casuale uniforme $r\in[0,1)$: la configurazione $x'$ viene accettata se $r<A(x\rightarrow x')$, altrimenti si mantiene la configurazione precedente.

Questa regola garantisce il bilancio dettagliato e quindi fa sì che p(x) sia la distribuzione stazionaria della catena di Markov. Dopo un numero sufficientemente elevato di iterazioni, il sistema raggiunge l'equilibrio e le configurazioni generate possono essere considerate campioni della distribuzione target. Il periodo iniziale, durante il quale il sistema evolve verso l'equilibrio, viene detto **tempo di equilibrazione** o **transitorio**.

### Applicazione al modello di Ising

Nel modello di Ising consideriamo un sistema di spin \(s_i=\pm1\) disposti su un reticolo. L'energia di una configurazione è

$$
E=-J\sum_{\langle i,j\rangle}s_i s_j,
$$

dove \(J\) è la costante di interazione e la somma è effettuata sulle coppie di primi vicini.

Nel nostro caso utilizziamo l'algoritmo di Metropolis per campionare la distribuzione canonica di Boltzmann

$$
P(\{s_i\})=\frac{1}{Z}e^{-\beta E(\{s_i\})},
$$

dove $\beta=1/(k_BT)$ e $Z$ è la funzione di partizione.

Ad ogni passo Monte Carlo viene scelto casualmente uno spin e se ne propone il flip,

$$
s_i\rightarrow -s_i.
$$

Si calcola quindi la variazione di energia

$$
\Delta E=E_{\mathrm{new}}-E_{\mathrm{old}}.
$$

La probabilità di accettazione diventa

$$
A=\min\left[1,e^{-\beta\Delta E}\right].
$$

Pertanto, se $\Delta E\leq0$, il flip viene sempre accettato, mentre se $\Delta E>0$ viene accettato solo con probabilità $e^{-\beta\Delta E}$.

Ripetendo questa procedura, la catena di Markov esplora lo spazio delle configurazioni dell'Ising e, una volta raggiunto l'equilibrio, permette di calcolare mediante medie Monte Carlo grandezze termodinamiche come energia, magnetizzazione, capacità termica e suscettività.
